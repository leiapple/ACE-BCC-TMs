## Screw dislocation core structure (differential displacement maps)
Relaxes a 135-atom screw-dislocation dipole for each element and plots the DD map of the core (Fig. 5(a)).

```python
# Import the function
from dislocation_analysis import calculate_dislocation_properties

# Calculator setup 
from pyace import PyACECalculator
```

```python


# Create calculator
calc_Mo = PyACECalculator('../ACE2025_potential/Mo/output_potential.yaml')
calc_Nb = PyACECalculator('../ACE2025_potential/Nb/output_potential.yaml')
calc_Ta = PyACECalculator('../ACE2025_potential/Ta/output_potential.yaml')
calc_V =  PyACECalculator('../ACE2025_potential/V/output_potential.yaml')
calc_W =  PyACECalculator('../ACE2025_potential/W/output_potential.yaml')

# Define parameters
elements = [ 'W', 'Mo', 'V', 'Ta', 'Nb']
a0_list = [ 3.185, 3.163, 2.99, 3.319, 3.307]
calculators = [calc_W, calc_Mo, calc_V, calc_Ta, calc_Nb]

# Run the analysis
results = calculate_dislocation_properties(
    elements=elements,
    a0_list=a0_list,
    calculators=calculators,
    fmax=0.005,
    optimizer='BFGS'
)

# Display results
print(results)

# Save to CSV
results.to_csv('dislocation_results.csv', index=False)
```

## Peierls barrier and core trajectory

For each element, the dipole is built in the initial and in the neighbouring Peierls valley and connected by a climbing-image NEB (11 images, FIRE, force tolerance 10⁻³ eV/Å, spring constant 0.1 eV/Å²). The barrier is given in meV/b (Fig. 6 and Table VII). The core trajectory is obtained from the stress variation of the NEB images (stress method), and the deviation angle α is measured from the initial easy core to the point of largest deviation (Fig. 5(b) and Table II).

```python
from dislocation_analysis import calculate_peierls_barrier_and_trajectory

peierls = calculate_peierls_barrier_and_trajectory(
    elements=elements,
    a0_list=a0_list,
    calculators=calculators,
    fmax=1e-3,
    spring_constant=0.1,
    Nreplica=11,
)
print(peierls)
peierls.to_csv('peierls_barrier_and_deviation_angle.csv', index=False)
```
